```python
#Import modules et fonctions
import requests
import pandas as pd
from datetime import timedelta
```

```python
#Recup url github
base_url="https://raw.githubusercontent.com/mikailaltundas/datasets-for-training/main/wind-power-dataset/"

#Recup chemin table bronze
Chemin_table_bronze="abfss://WindPowerAnalytic_Dev@onelake.dfs.fabric.microsoft.com/LH_Wind_Power_Bronze.Lakehouse/Tables/dbo/wind_power"
#display(base_url)
```

```python
#Lire la table wind_power
df_spark=spark.read.format("delta").load(Chemin_table_bronze)
df_pandas=df_spark.toPandas()
#display(df_pandas)
```

```python
#Recup date plus recente
Date_recente=pd.to_datetime(df_pandas["date"],format="%Y%m%d").max()
#display(Date_recente)
Date_suivante=(Date_recente+timedelta(days=1)).strftime("%Y%m%d")
#display(Date_suivante)
```

```python
#Créer l'url et récupére le fichier csv du jour
file_url=f"{base_url}{Date_suivante}_wind_power_data.csv"
#Lire les données contenu dans l'url
df_pendas_nouveau=pd.read_csv(file_url)
df_pendas_nouveau["date"]=pd.to_datetime(df_pendas_nouveau["date"])
#display(df_pendas_nouveau)
```

```python
df_spark_nouveau=spark.createDataFrame(df_pendas_nouveau,schema=df_spark.schema)
```

```python
df_spark_nouveau.write.format("delta").mode("append").save(Chemin_table_bronze)
```
